# 01 データ確認

Football-Data.co.uk形式のCSVを読み込み、件数・欠損・結果割合を確認します。まずCSVを `data/raw/football_data_co_uk` に入れてください。

```python
from pathlib import Path
import sys
ROOT=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
sys.path.insert(0,str(ROOT))
from ml.data_loader import load_raw_matches, save_clean_matches
```

```python
df=load_raw_matches()
print('試合数:',len(df))
print('期間:',df.Date.min(),'〜',df.Date.max())
df.head()
```

```python
print(df.FTR.value_counts(normalize=True).rename({'H':'ホーム勝ち','D':'引き分け','A':'アウェイ勝ち'}))
print('\n欠損率上位')
display(df.isna().mean().sort_values(ascending=False).head(20))
```

```python
path=save_clean_matches(df)
print('保存:',path)
```
