各試合より前の情報だけを使って直近成績・得失点・休養日数などを生成します。

```python
from pathlib import Path
import sys
ROOT=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
sys.path.insert(0,str(ROOT))
from ml.data_loader import load_raw_matches
from ml.features import build_features,feature_columns
from ml.config import PROCESSED_DIR
```

```python
raw=load_raw_matches()
features=build_features(raw,window=5)
print(features.shape)
display(features.head())
```

```python
display(features[feature_columns(features)].describe().T)
print(features.target.value_counts(normalize=True).sort_index())
```
